# Cluster Backtest

Synthetic parameter exploration only. This does not reveal tuned values; paper trade live before taking risk.

```python
import json
from pathlib import Path

from clusterdetect import ClusterDetector, PaperTrader

swaps = json.loads(Path("examples/cluster_backtest/sample_swaps.json").read_text())
scores = {
    "DemoWallet_Alpha_not_a_real_address": 3,
    "DemoWallet_Beta_not_a_real_address": 2,
    "DemoWallet_Gamma_not_a_real_address": 1,
    "DemoWallet_Delta_not_a_real_address": 1,
}
rows = []
for min_wallets in [2, 3, 4]:
    for window in [5, 15, 30]:
        detector = ClusterDetector(
            min_wallets=min_wallets, window_minutes=window, min_total_score=6
        )
        clusters = detector.detect(swaps, scores)
        rows.append({"min_wallets": min_wallets, "window": window, "clusters": len(clusters)})
rows
```
